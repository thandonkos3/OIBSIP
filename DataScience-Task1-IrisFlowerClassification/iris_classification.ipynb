## 6. Data Loading

The Iris dataset will be loaded using the Scikit-learn library. It is a built-in dataset containing 150 flower observations belonging to three different iris species.

Each observation contains four numerical features:

- Sepal length
- Sepal width
- Petal length
- Petal width

The target variable represents the species of the iris flower.

The dataset will be converted into a Pandas DataFrame to make data exploration and analysis easier.


```python

# Load the Iris dataset
iris = load_iris()

# Create a DataFrame containing the flower measurements
df = pd.DataFrame(
    data=iris.data,
    columns=iris.feature_names
)

# Add the target species column
df['species'] = iris.target

# Convert numerical target values into species names
df['species'] = df['species'].apply(
    lambda x: iris.target_names[x]
)

# Display the first five records
df.head()
```


## 11. Feature Selection

Feature selection is the process of identifying the most relevant variables that contribute to accurate machine learning predictions.

The Iris dataset contains four numerical features:

- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

Petal length and petal width are expected to be particularly useful for distinguishing between the three iris species because their measurements show noticeable differences across the species.

The relationship between these features will be examined using correlation analysis and visualisations.

The selected features will be used during the machine learning model development stage.


```python

# Examine the correlation between numerical features
feature_correlation = df.drop(
    columns=['species']
).corr()

# Display correlations with petal length
print("Correlation with Petal Length:")
print(
    feature_correlation['petal length (cm)']
    .sort_values(ascending=False)
)

# Display correlations with petal width
print("\nCorrelation with Petal Width:")
print(
    feature_correlation['petal width (cm)']
    .sort_values(ascending=False)
)
```


## 12. Data Preprocessing

Data preprocessing prepares the dataset for machine learning.

The following steps will be performed:

1. Separate the independent variables (features) from the dependent variable (target).
2. Assign the four flower measurements to the feature variable X.
3. Assign the species labels to the target variable y.
4. Split the dataset into training and testing sets.
5. Use an 80:20 split, allocating 80% of the data for training and 20% for testing.

This process will help ensure that the machine learning models are trained and evaluated using separate data.


```python

# Separate features (X) and target variable (y)
X = df.drop(columns=['species'])
y = df['species']

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42, 
    stratify=y
)  

# Display the dataset split sizes
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```


## 13. Machine Learning Model Development

In this stage, different supervised machine learning classification algorithms will be trained using the Iris dataset.

The following algorithms will be evaluated:

1. Logistic Regression  
2. K-Nearest Neighbors (KNN)
3. Decision Tree Classifier
4. Random Forest Classifier

Each model will learn patterns from the training dataset and use those patterns to predict the species of flowers in the testing dataset.

The models will then be compared to identify which algorithm provides the best classification performance.


```python
print("Jupyter kernel is working correctly!") 
```

```python

# MODEL 1: LOGISTIC REGRESSION

# Create a Logistic Regression pipeline
logistic_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)

# Train the model using the training dataset
logistic_model.fit(X_train, y_train)

print("Logistic Regression model training completed.")
```

```python

# MODEL 2: K-NEAREST NEIGHBORS (KNN)

# Create a KNN pipeline
knn_model = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)

# Train the model using the training dataset
knn_model.fit(X_train, y_train)

print("KNN model training completed.")
```

```python

# MODEL 3: DECISION TREE CLASSIFIER

# Create the Decision Tree model
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model using the training dataset
decision_tree_model.fit(X_train, y_train)

print("Decision Tree model training completed.")
```

```python

# MODEL 4: RANDOM FOREST CLASSIFIER

# Create the Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model using the training dataset
random_forest_model.fit(X_train, y_train)

print("Random Forest model training completed.")
```

## 14. Model Evaluation

Model evaluation is the process of measuring how well a trained machine learning model performs on unseen data.

In this project, the four trained classification models will be evaluated using the testing dataset:

- Logistic Regression
- K-Nearest Neighbors (KNN)
- Decision Tree
- Random Forest

The accuracy score will be used to measure the percentage of correctly classified iris flowers. The results will then be compared to identify which model performs best.

```python
# MODEL EVALUATION

models = {
    'Logistic Regression': logistic_model,
    'K-Nearest Neighbors': knn_model,
    'Decision Tree': decision_tree_model,
    'Random Forest': random_forest_model
}

results = []

for model_name, model in models.items():
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)

    results.append({
        'Model': model_name,
        'Accuracy': accuracy
    })

results_df = pd.DataFrame(results)

display(
    results_df.sort_values(
        by='Accuracy',
        ascending=False
    )
)
```

## 15. Classification Report

A classification report provides a detailed evaluation of each machine learning model using precision, recall, and F1-score.

These metrics help determine how effectively each model identifies the three iris flower species: Setosa, Versicolor, and Virginica.

The classification reports will be generated for all four trained models and compared to assess their classification performance.

```python
# CLASSIFICATION REPORTS

for model_name, model in models.items():
    print(f"\n{'=' * 50}")
    print(f"Classification Report: {model_name}")
    print('=' * 50)

    y_pred = model.predict(X_test)

    print(
        classification_report(
            y_test,
            y_pred
        )
    )
```
